```python
from pathlib import Path

from huggingface_hub import snapshot_download

grand_tour_folder = Path("/datahome/lmertens/grand_tour_dataset")


# allow_patterns = ["2024-10-01-11-29-55/*"]
# ignore_patterns = [
#     "*lidar*",
#     "*zed2i_depth_confidence_image.tar",
#     "*hesai*",
#     "*hdr*",
#     "*alphasense*",
#     "*livox*",
#     "*depth_camera*",
# ]


mission = "2024-11-11-12-07-40"

topics = [
    "anymal_command_twist",
    "cpt7_ie_tc_odometry",
    "cpt7_ie_tc_tf",
    "cpt7_imu",
    "gnss_raw_cpt7_ie_tc",
    "navsatfix_cpt7_ie_tc",
    "prism_position",
    "stim320_imu",
    "tf",
    "zed2i_left_images",
]

allow_patterns = [f"{mission}/*.yaml", f"{mission}/*.zgroup"]
allow_patterns += [f"{mission}/*/{topic}.*" for topic in topics]

ignore_patterns = []


dry_run_file_info = snapshot_download(
    repo_id="leggedrobotics/grand_tour_dataset",
    allow_patterns=allow_patterns,
    # ignore_patterns=ignore_patterns,
    repo_type="dataset",
    dry_run=True,
)


newly_downloaded_size = sum(
    file_info.file_size for file_info in dry_run_file_info if file_info.will_download
)
total_size = sum(file_info.file_size for file_info in dry_run_file_info)
total_num_files = len(dry_run_file_info)

print(f"Total size of files to be downloaded: {newly_downloaded_size / (1e6):.2f} MB")
print(f"Total size of all files (including cached): {total_size / (1e6):.2f} MB")
print(f"Total number of files: {total_num_files}")
```

```python
included_files = [
    f"{file_info.filename} - {file_info.file_size / (1e6):.2f} MB"
    for file_info in sorted(dry_run_file_info, key=lambda x: x.file_size, reverse=True)
]

print(*included_files, sep="\n")
```
