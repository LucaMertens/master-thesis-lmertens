```python
from pathlib import Path

import kleinkram
from data_labeling_pipeline.grandtour_missions import get_mission_by_shortname

grand_tour_folder = Path("/datahome/lmertens/grand_tour_dataset")

# mission_shortname = "RIV-1"
mission_shortname = "ETH-1"
mission_info = get_mission_by_shortname(mission_shortname)
mission_timestamp = mission_info["timestamp"]
mission_name = f"release_{mission_timestamp}"


download_folder = grand_tour_folder / "ros1" / mission_shortname
download_folder.mkdir(parents=True, exist_ok=True)

patterns = [
    "*anymal_command*",
    "*cpt7_ie_tc*",
    "*tf_minimal*",
    "*zed2i_images*",
]
```

```python
missions = kleinkram.list_missions(
    project_names=["GrandTourDataset"], mission_names=[mission_name]
)
if not missions:
    print(f"Mission {mission_timestamp} not found in Kleinkram.")
else:
    m = missions[0]
    files = kleinkram.list_files(mission_ids=[m.id], file_names=patterns)
    print(
        f"Found {len(files)} files to download for mission {mission_shortname} ({mission_name})"
    )
    for f in files:
        print(f"- {f.name} ({f.size / 1e6:.2f} MB)")
```

```python
if missions and files:
    print("Downloading dataset from Kleinkram...")
    kleinkram.download(
        file_ids=[f.id for f in files],
        dest=str(download_folder),
        verbose=True,
        overwrite=True,
    )
    print("Download complete.")
```
